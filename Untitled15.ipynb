Bag of words matrix

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "i love pizza",
    "pizza is the best",
    "i love pasta",
    "pasta is graet"
]

vectorizer = CountVectorizer()

x= vectorizer.fit_transform(documents)

print("voccabualry: ",vectorizer.get_feature_names_out())
print("\n bag of matrix:\n",x.toarray())
```

Tf idf

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "i love pizza",
    "pizza is the best",
    "i love pasta",
    "pasta is great"
]

vectorizer = TfidfVectorizer()

x = vectorizer.fit_transform(documents)

print("voccabualry:",vectorizer.get_feature_names_out())
print("\nBow Matrix:",x.toarray())
```
