```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("train.txt",sep = ';',header=None,names=['text','emotion'])
```

```python
emotion_numbers = {}
```

```python
unique_emotions = df['emotion'].unique()
emotion_numbers = {}
i= 0
for emo in unique_emotions:
  emotion_numbers[emo] = i
  i+=1

df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
df['text'] = df['text'].apply(lambda x: x.lower())
```

```python
import string
def remove_punctuation(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punctuation)
```

```python
def remove_number(txt):
  new=""
  for i  in txt:
    if not i.isdigit():
      new = new+i
  return new

df['text'] = df['text'].apply(remove_punctuation)
```

remove url links


remove emojis and special characters

```python
from os import device_encoding
def remove_emojis(txt):
  new= ""
  for i in txt:
    if i.isascii():
      new+=i
  return new
df['text'] = df['text'].apply(remove_emojis)
```

remove stopwards : doing all the tasks are easy for removing the stop words like 'is','the','are' etc we have to use a library and that is NLTK

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import  word_tokenize
```

```python
nltk.download('punkt_tab')
```

```python
nltk.download('punkt') # use for tokenization
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
  words = word_tokenize(txt)
  cleaned = []
  for i in words:
    if i not in stop_words:
      cleaned.append(i)

  return ''.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(df['text'],df['emotion'],test_size =0.2,random_state =42)
```

text vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer,CountVectorizer
```

```python
bow_vectorizer = CountVectorizer()
x_train_bow = bow_vectorizer.fit_transform(x_train)
x_test_bow = bow_vectorizer.transform(x_test)
```

using statistical approches of machine learning

for text classification multinomial naive bayes should use for it

```python
from sklearn.naive_bayes import  MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb_model = MultinomialNB()
nb_model.fit(x_train_bow,y_train)
```

```python
y_pred = nb_model.predict(x_test_bow)
```

```python
print(accuracy_score(y_test,y_pred))
```
